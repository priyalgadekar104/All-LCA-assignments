```python
#creating a dictionary
student={
    "name": "abc",
    "age": 18,
"branch": "cse core"
}
print(student)
```

```python
# adding a new element
student["college"]="abc college"
print("\nAfter adding an element:")
print(student)
```

```python
#updating an element
student["age"]=19
print("\nAfter updating age:")
print(student)
```

```python
#removing an element
student.pop("branch")
print("\nAfter removing branch:")
print(student)
```

```python
#creating a tuple
numbers=(10,20,30,40,50)
print(numbers)
```

```python
#accessing an element
print("\nFirst element:", numbers[0])
```

```python
#slicing
print( numbers[1:4])
```

```python
#finding length
print(len(numbers))
```

```python
#counting an element
print(numbers.count(20))
```

```python
#to add an element, convert tuple to list
temp=list(numbers)
temp.append(60)
numbers=tuple(temp)
print(numbers)
```

```python
#removing an element
temp=list(numbers)
temp.remove(30)
numbers=tuple(temp)
print(numbers)
```
